```python
# 가장 먼저 실습을 위해 필요한 랭체인 도구들을 임포트함.
import os
import urllib.request
from langchain_openai import OpenAIEmbeddings
from langchain_experimental.text_splitter import SemanticChunker

os.environ["OPENAI_API_KEY"] = ""

# OpenAI의 Embedding API는 기본적으로 유료이므로 너무 긴 텍스트를 실습하지 않기를 권장함.
urllib.request.urlretrieve("https://github.com/ai-agent-kr/agent-tutorial/blob/main/Ch02/test.txt", filename="test.txt")
with open("test.txt", encoding="utf-8") as f:
    file = f.read()

#print('텍스트의 길이:', len(file))

# 이제 SemanticChunker 로 텍스트를 분할해 봄.
# 객체를 만드는 것과 분할 하는 방법은 RecursiveCharacterTextSplitter 와 거의 동일함.
# SemanticChunker()를 이용해 텍스틀르 분할하는 text_splitter객체를 만듬.
# 이때 OpenAIEmbeddings()를 전달해 OpenAI의 Embedding API를 사용한다는 것읅 명시함.
text_splitter = SemanticChunker(OpenAIEmbeddings())
texts = text_splitter.create_documents([file,])
#print('분할된 청크의 수', len(texts))

texts[3]

# SemanticChunker 가 문서를 분할하는 세가지 방법
# - 백분위수 (기본값)
# - 표준편차
# - 사분위수

# 우선 이 세가지 방식 은 모두 코사인 거리라는 개념을 사용함.
#코사인 거리는 두 문장간의 의미적 차이를 나타내는 척도임.
# 코사인 거리는 '코사인 유사도'에서 설명한 두 임베딩 벡터의 유사도인 코사인 유사도로부터 계산 할 수 있음.
# 두 벡터간의 코사인 유사도를 계산하고 1 에서 빼면 코사인 거리임.
# 코사인 거리 = 1 - 코사인 유사도 (0~2) 세가지 방식을 적용하기 위해 각 문장은 임베딩 벡터로 변환되며, 인접한 문장 쌍 사이의 코사인 거리를 계산함.
# 이 때 코사인 거리는 값의 범위는 0 ~ 2 사이이며 0에 가까울수록 유사하고 2에 가까울수록 다름.
```

```python
import os
import urllib.request
from langchain_openai import OpenAIEmbeddings
from langchain_experimental.text_splitter import SemanticChunker

os.environ["OPENAI_API_KEY"] = ""

urllib.request.urlretrieve("https://github.com/ai-agent-kr/agent-tutorial/blob/main/Ch02/test.txt", filename="test.txt")
with open("test.txt", encoding="utf-8") as f:
    file = f.read()

# 1. 백분위수 방법
# 백분위수 방법은 SematicChunker 가 기본으로 사용하는 방식임.
# 백분위수 방식을 사용할 때는 SemanticChunker 를 이용해 text_splitter 를 선언할 때 기준이 되는 값에 해당하는 breakpoint_threshold_amount 의 값을 임의로 설정할 수 있음.
# breakpoint_threshold_amount 의 기본값은 95임
# 따라서 다음 코드는 위의 실습에서 사용한 코드와 동일한 결과를 가짐.
#text_splitter = SemanticChunker(
#    OpenAIEmbeddings(),
#    breakpoint_threshold_type = "percentile",
#    breakpoint_threshold_amount = 95
#)

#texts = text_splitter.create_documents([file])
#print('분할된 청크의 수 : ', len(texts))

# 원리
# 모든 연속된 문장 쌍 사이의 코사인 거리를 계산함.
# 이 코사인 거리들을 크기 순으로 정렬함.
# 예를 들어, breakpoint_threshold_amount 의 값을 95로 설정한 경우, 95번째 백분위수에 해당하는 코사인 거리를 찾음.
# 즉 이렇게 찾은 전체 거리 중 95% 가 이 값보다 적음.
# 이 값을 기준으로 삼아 이 보다 큰 코사인 거리를 가진 지점에서 텍스트를 나눔.
# 예시
# 100개의 문장이 있다면 99개의 코사인 거리 값이 생김.
# 이 99개의 값을 정렬하고 95번째로 큰 값을 찾음.
# 이 값보다 큰 코사인 거리를 가진 곳에서만 텍스트를 나눔.
# 장점.
# 극단적으로 큰 의미적 차이가 있는 곳에서만 텍스트를 나누므로 주요 주제가 바뀌는 곳을 잘 찾을 수 있음.

# 2. 표준 편차 방식
# 표준 편차 방식을 사용하는 경우에는 breakpoint_threshold_type 값으로 standard_deviation 을 사용하면 됨.
# 이때 breakpoint_threshold_amount 의 기본값은 3임
#text_splitter = SemanticChunker(
#    OpenAIEmbeddings(),
#    breakpoint_threshold_type="standard_deviation",
#    breakpoint_threshold_amount = 3
#)

#texts = text_splitter.create_documents([file])
#print('분할된 청크의 수 : ', len(texts))

# 기본값
# breakpoint_threshold_amount = 3
# 원리
# 모든 연속된 문장 쌍 사이의 코사인 거리를 계산함.
# 이 코사인 거리들의 평균과 표준편차를 계산함.
# 예를 들어 breakpoint_threshold_amount 값을 3으로 할 경우 (평균 + 3 * 표준편차)를 기준값으로 삼음.
# 예시
# 코사인 거리의 평균이 0.5 이고, 표준편차가 0.1이라면
# 기준값은 0.5 + 3 * 0.1 = 0.8 이 됨.
# 0.8 보다 큰 코사인 거리를 가진 곳에서 텍스트를 나눔.
# 장점.
# 텍스트 전체의 의미적 흐름을 고려해 평균적인 문장 간차이보다 훨씬 큰 변화가 있는 지점을 찾아냄.
# 이는 새로운 주체의 시작, 논점의 전환, 또는 이야기의 큰 전환점 등을 효과적으로 감지할 수 있게 해 줌.

# 3. 사분위수 방식.
# 사분위수(Interquatile)방식을 사용해 봄.
# 사분위수 방식을 사용하는 경우에는 breakpoint_threshold_type 의 값으로 Interquartile 을 사용하면 됨.
# 이때 breakpoint_threshold_amount의 기본값을 1.5임.

text_splitter = SemanticChunker(
    OpenAIEmbeddings(),
    breakpoint_threshold_type="interquartile",
    breakpoint_threshold_amount=1.5
)
texts = text_splitter.create_documents([file])
print('분할된 청크의 수 : ', len(texts))

# 기본값
# breakpoint_threshold_amount = 1.5
# 원리
# 모든 연속된 문장 쌍 사이의 코사인 거리를 계산함.
# 이 거리들의 1사분위수(Q1, 25% 지점), 3사분위수(Q3, 75& 지점)을 찾음
# 사분위수 범위(IQR) = Q3 - Q1 를 계산함.
# 거리들의 평균에 (breakpoint_threshold_amount * IQR)을 더한 값으로 기준값으로 삼음.
# 이 기준값보다 큰 거리를 가진 지점에서 텍스트를 나눔.
# 예시
# Q1이 0.3, Q3가 0.7, 평균이 0.5, breakpoint_threshold_amount = 1.5라면
# IQR = 0.7 - 0.3 = 0.4임
# 기준값은 0.5 + (1.5 * 0.4) = 1.1 이 됨.
# 1.1 보다 큰 거리를 가진 곳에서 텍스트를 나눔.

# 장점
# 텍스트의 전반적인 구조를 고려하면서도 지나치게 민감하지 않게 분할점을 찾음
# 특히 긴 ㅁ누서에서 주요 섹션의 경계를 식별하는데 효과적임
# 또한 이상치(극단적으로 다른 부분)에 덜 민감해서 텍스트의 전반적인 흐름을 해치치 않으면서도 중요한 주제 전환을 포착할 수 있음.

# 실제로 문서를 어떻게 자르느냐가 RAG 챗봇 성능에 영향을 미치지 중요한 요소 중 하나임.
# RecursiveCharacterTextSplitter의 결과가 만족스럽지 않다면 대안으로 고려해 보기 바람.
```
