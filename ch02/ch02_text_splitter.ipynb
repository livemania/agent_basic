```python
# 실습에 필요한 랭체인 도구들을 임포트함.
import urllib.request
from langchain_text_splitters import RecursiveCharacterTextSplitter

# 실습을 위해 굉장히 긴 길이를 가진 텍스트를 다운로드 하고 길이를 출력함.
urllib.request.urlretrieve("https://raw.githubusercontent.com/lovit/soynlp/master/tutorials/2016-10-20.txt", filename="2016-10-20.txt")

with open("2016-10-20.txt", encoding="utf-8") as f:
    file = f.read()

print("텍스트의 길이:", len(file))

# 텍스트의 길이가 너무 길어, 일반적으로 ChatGPT 같은 언어 모델에 한번에 넣을 수 없는 긴 길이의 텍스트임.
# 이제 이 텍스트를 ChatGPT  같은 언어 모델이 처리할 수 있는 적당한 길이로 분할 해 봄.
# RecursiveCharacterTextSplitter()를 이용해 텍스트를 분할하는 text_splitter 객체를 만듬.
# 이때 chunk_size 의 값을 500으로 지정하면 앞으로 text_splitter 로 텍스트를 분할 할때 각 분할된 청크의 길이가 500을 결코 넘지 않음.
# chunk_overlap 은 텍스트를 분할할 때 각 청크가 내용을 얼만큼 겹치게 할 것인지를 정하는 값으로 0을 지정하면 각 청크의 내용이 겹치지 않음.
text_splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=0)

# 이렇게 선언된 text_splitter로 파이썬 문자열을 분할할때는 create_documents()를 사용함.
texts = text_splitter.create_documents([file])
print('분할된 청크의 수:', len(texts))
```

```python
# 출력 결과를 보면 Document(metadata={}, page_content="") 형식을 가짐.
# 이 형태는 랭체인을 이용해 텍스트를 다수의 청크로 분할 했을 때 갖게 되는 형태로, 앞으로 랭체인 실습을 하면서 자주 보게 될 형태이므로 기억해 둠.
# 이때 원문에 접근하려면 각 청크에 .page_content 를 붙여서 출력하면 됨.
texts[1].page_content
```

```python
print('청크 1의 길이', len(texts[1].page_content))
print('청크 2의 길이', len(texts[2].page_content))
```
